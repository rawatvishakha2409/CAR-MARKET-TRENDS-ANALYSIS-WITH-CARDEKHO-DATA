```python
#Pie chart
Fuel_count = data.Fuel_Type.value_counts()

plt.figure(figsize=(8, 8))
plt.pie(
    Fuel_count,
    labels=Fuel_count.index,
    autopct="%1.1f%%")
plt.title("Cars by Fuel Type")

plt.show()
```

```python
#Type of Seller
print(data.Seller_Type.value_counts())
plt.figure(figsize=(8, 8))
plt.pie(data.Seller_Type.value_counts(), labels=data.Seller_Type.value_counts().index, autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Seller Types')
plt.ylabel('') # Hide the default 'y' label for pie charts
plt.show()
```

```python
#Bar graph
Fuel_Type_counts = data.Fuel_Type.value_counts()

plt.figure(figsize=(8, 6))
sns.barplot(x=Fuel_Type_counts.index, y=Fuel_Type_counts.values, palette='viridis')
plt.title('Count of Cars by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Cars')
plt.show()
```

```python
# Create Bar Chart
avg_price = data.groupby('Transmission')['Selling_Price'].mean()
plt.figure(figsize=(8, 6))

plt.bar(avg_price.index, avg_price.values)

plt.title("Average Selling Price by Transmission Type")
plt.xlabel("Transmission Type")
plt.ylabel("Average Selling Price")

plt.show()
```

```python
#Line Chart
year_counts = data.Year.value_counts().sort_index()

plt.figure(figsize=(10, 6))
plt.plot(year_counts.index, year_counts.values)
plt.title('Number of Cars by Year')
plt.xlabel('Year')
plt.ylabel('Number of Cars')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

```python
# Bar chart for car count per year
plt.figure(figsize=(12, 6))

sns.countplot(x='Year', data=data, palette='coolwarm')
plt.title('Number of Cars per Year')
plt.xlabel('Year')
plt.ylabel('Number of Cars')
plt.xticks(rotation=45)

plt.show()
```

```python
# Create Histogram
plt.figure(figsize=(10, 6))

plt.hist(data["Selling_Price"], bins=20, edgecolor="black")
plt.title("Distribution of Car Selling Prices")
plt.xlabel("Selling Price")
plt.ylabel("Number of Cars")

plt.show()
```

```python
# Create Scatter Plot
plt.figure(figsize=(10, 6))

plt.scatter(data["Year"], data["Selling_Price"])
plt.title("Year vs Selling Price")
plt.xlabel("Year")
plt.ylabel("Selling Price")

plt.show()
```
